```python

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import parameters as _par
from conditions import (print_regime_summary, regime_map, n_star,
                        slope_R, right_branch_type, MODES)
```

```python

# ── parametri ──────────────────────────────────────────────────────────────
kw = dict(
    A_p_bar  = _par.A_p_bar,
    A_up_bar = _par.A_up_bar,
    mu       = _par.mu,
    r        = _par.r,
    alpha    = _par.alpha,
    xF       = _par.xF,
    Ybar     = _par.Ybar,
)

# ── 1. stampa diagnostica ──────────────────────────────────────────────────
print_regime_summary(**kw)

# ── 2. plot: regime al variare di n_p per i tre mode ──────────────────────
PALETTE = {
    "unique_poor_attractor"        : "steelblue",
    "poor_and_rich_attractor"      : "seagreen",
    "poor_attractor_and_explosion" : "orange",
    "unique_rich_attractor"        : "gold",
    "indefinite_growth"            : "firebrick",
    "undetermined"                 : "gray",
}

fig, axes = plt.subplots(3, 1, figsize=(11, 4), sharex=True)

for ax, mode in zip(axes, MODES):
    n_vals, regimes = regime_map(mode, n_grid=300, **kw)
    colors = [PALETTE.get(r, "gray") for r in regimes]

    ax.scatter(n_vals, np.zeros_like(n_vals),
               c=colors, s=60, marker="|", linewidths=1.5)

    ns = n_star(mode, kw["A_p_bar"],
                mu=kw["mu"], r=kw["r"], alpha=kw["alpha"])
    if 0 < ns < 1:
        ax.axvline(ns, color="black", linestyle="--", linewidth=1.2,
                   label=f"n* = {ns:.3f}")
        ax.legend(fontsize=8, loc="upper left")

    ax.set_ylabel(mode, fontsize=9)
    ax.set_yticks([])

axes[-1].set_xlabel("$n^P$", fontsize=11)
fig.suptitle("Regime al variare di $n^P$ — confronto tra i tre mode", fontsize=11)

legend_patches = [mpatches.Patch(color=v, label=k) for k, v in PALETTE.items()
                  if k != "undetermined"]
fig.legend(handles=legend_patches, loc="lower center",
           ncol=3, fontsize=8, bbox_to_anchor=(0.5, -0.05))
plt.tight_layout()
plt.savefig("/mnt/user-data/outputs/regime_map.png", dpi=150, bbox_inches="tight")
plt.show()
print("Plot salvato in outputs/regime_map.png")

# ── 3. plot: a_R(n_p) per i tre mode ──────────────────────────────────────
fig2, ax2 = plt.subplots(figsize=(8, 4))
n_vals = np.linspace(0, 1, 200)
colors_mode = {"expected": "steelblue", "noshock": "seagreen", "shock": "firebrick"}

for mode in MODES:
    slopes = [slope_R(n, mode, kw["A_p_bar"],
                      mu=kw["mu"], r=kw["r"], alpha=kw["alpha"])
              for n in n_vals]
    ax2.plot(n_vals, slopes, label=mode, color=colors_mode[mode], linewidth=2)
    ns = n_star(mode, kw["A_p_bar"],
                mu=kw["mu"], r=kw["r"], alpha=kw["alpha"])
    if 0 < ns < 1:
        ax2.axvline(ns, color=colors_mode[mode],
                    linestyle=":", linewidth=1.2, alpha=0.7)

ax2.axhline(1, color="black", linestyle="--", linewidth=1, label="a_R = 1  (n*)")
ax2.set_xlabel("$n^P$", fontsize=11)
ax2.set_ylabel("$a_R(n^P)$", fontsize=11)
ax2.set_title("Slope del ramo destro al variare di $n^P$", fontsize=11)
ax2.legend(fontsize=9)
ax2.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("/mnt/user-data/outputs/slope_R.png", dpi=150, bbox_inches="tight")
plt.show()
print("Plot salvato in outputs/slope_R.png")
```
